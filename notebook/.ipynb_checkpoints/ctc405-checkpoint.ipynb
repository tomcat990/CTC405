# Help and Documentation

**The `help` command and the Documentation are your best tools. for example:**


```octave
help("cast")
```

## Octave as a Calculator

- Data types: `int`, `float`, `long`, `string`, `char`, ...
- Simple operations.
- `Inf`, `pi`, `intmax`, `realmax`, ...
- Defining variables (be aware of keywords):
  - `iskeyword('foo')`
  - `isvarname('foo')`
- Use `,` to combine several statements.
- `clear foo`, `clear all`, `clc`, ...
- `whos`, `who`.
- Complex numbers: `i`, `j`, `real`, `imag`, `abs`, `conj`.
- `sin`, `sind`, `sinh`, `asin`, ...
- `sqrt`, `nthroot`, `factorial`, `log`, `log10`, `gcd`, `lcm`, `rem`, `mod`, ...
- `round`, `fix`, `floor`, `ceil`.

```octave
cast(2048,"int8"),cast(2048,"uint8"),cast(2048,"int16")
```

```octave
sin(30),sind(30),asin(0.5),asind(0.5)
```

## Vectors & Matrices

- Defining vectors: simple version and `n:step:m`.
- `ones`, `zeros`, `eye`, ...
- `linspace`, `logspace`.
- `rand`, `randn`, `randi`, ...
- Extraction and deletion from vectors and matrices.
- `inv`, `det`, transpose, `max`, `mean`, `sort`, `diag`, `sum`, `prod`.
- Addition, subtraction, division and multiplication of elements.

**Matrix division**

$$
\begin{aligned}
XA=B &\quad \rightarrow \quad X=B/A \\[8pt]
AX=B &\quad \rightarrow \quad X=A^{-1}B
\end{aligned}
$$

```octave
eye(3),zeros(3),ones(3)
```

```octave
linspace(1,100,10),disp('---------------------------------------------------'),logspace(1,5,10)
```

```octave
help rand
```

```octave
mat1 = rand(3),inv(mat1),det(mat1),mat1'
```
